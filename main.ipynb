```python
from pathlib import Path

folder = Path("data")
all_content = "\n\n---\n\n".join(
    file_path.read_text(encoding="utf-8") for file_path in sorted(folder.glob("*.md"))
)

print(len(all_content))
```

```python

from langchain_text_splitters import MarkdownHeaderTextSplitter

headers_to_split_on = [
    ("#", "Header 1"),
    ("##", "Header 2"),
    ("###", "Header 3"),
]


markdown_splitter = MarkdownHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
md_header_splits = markdown_splitter.split_text(all_content)
```

```python
for doc in md_header_splits:
    h1 = doc.metadata.get("Header 1")
    h2 = doc.metadata.get("Header 2")
    h3 = doc.metadata.get("Header 3")
    print(f"H1: {h1} | H2: {h2} | H3: {h3} -> Text: {doc.page_content}")
```
